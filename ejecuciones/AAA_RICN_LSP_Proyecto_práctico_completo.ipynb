````python
from __future__ import division

from PIL import Image
import numpy as np
import gym

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Activation, Flatten, Convolution2D, Permute

if IN_COLAB:
  from tensorflow.keras.optimizers.legacy import Adam
else:
  from tensorflow.keras.optimizers import Adam

import tensorflow.keras.backend as K

from rl.agents.dqn import DQNAgent
from rl.policy import LinearAnnealedPolicy, BoltzmannQPolicy, EpsGreedyQPolicy
from rl.memory import SequentialMemory
from rl.core import Processor
from rl.callbacks import FileLogger, ModelIntervalCheckpoint
````

#### Configuración base

````python
# Descomentar y adaptar a cada entorno según corresponda
INPUT_SHAPE = (84, 84)
WINDOW_LENGTH = 4

env_name = 'Enduro-v0'
env = gym.make(env_name)

np.random.seed(123)
env.seed(123)
nb_actions = env.action_space.n

# Umbral objetivo: superar el primer día de carrera (200 coches)
# de forma sostenida durante los 100 episodios de test
REWARD_THRESHOLD = 200


# --- Definición de las 3 propuestas a comparar ---
experiments = {
    'baseline_dqn':   dict(double_dqn=False, dueling=False),
    'double_dqn':     dict(double_dqn=True,  dueling=False),
    'dueling_dqn':    dict(double_dqn=False, dueling=True),
}

NB_STEPS_TRAIN = 1750000  # ajustar según tiempo/recursos disponibles
````

````python
class AtariProcessor(Processor):
    def process_observation(self, observation):
        assert observation.ndim == 3  # (height, width, channel)
        img = Image.fromarray(observation)
        img = img.resize(INPUT_SHAPE).convert('L')
        processed_observation = np.array(img)
        assert processed_observation.shape == INPUT_SHAPE
        return processed_observation.astype('uint8')

    def process_state_batch(self, batch):
        processed_batch = batch.astype('float32') / 255.
        return processed_batch

    def process_reward(self, reward):
        return np.clip(reward, -1., 1.)
````

1. Implementación de la red neuronal

````python
from tensorflow.keras.layers import Layer, Input
from tensorflow.keras.models import Model

if IN_COLAB:
    # Solo necesario en Colab con TF 2.18 / Keras 3, para permitir
    # la deserialización del Lambda de combinación del Dueling DQN
    from tf_keras.src.saving import serialization_lib
    serialization_lib.in_safe_mode = lambda: False

class DuelingCombine(Layer):
    def call(self, inputs):
        value, advantage = inputs
        return value + (advantage - K.mean(advantage, axis=1, keepdims=True))

def build_model(window_length, input_shape, nb_actions, dueling=False):
    input_shape_full = (window_length,) + input_shape
    frame = Input(shape=input_shape_full)
    x = Permute((2, 3, 1))(frame)

    x = Convolution2D(32, (8, 8), strides=(4, 4), activation='relu')(x)
    x = Convolution2D(64, (4, 4), strides=(2, 2), activation='relu')(x)
    x = Convolution2D(64, (3, 3), strides=(1, 1), activation='relu')(x)
    x = Flatten()(x)
    x = Dense(512, activation='relu')(x)

    if dueling:
        value = Dense(1, activation='linear')(x)
        advantage = Dense(nb_actions, activation='linear')(x)
        outputs = DuelingCombine()([value, advantage])
    else:
        outputs = Dense(nb_actions, activation='linear')(x)

    model = Model(inputs=frame, outputs=outputs)
    print(model.summary())
    return model
````

La arquitectura convolucional utilizada es muy similar a la propuesta originalmente por DeepMind para Atari.

Las capas convolucionales permiten extraer automáticamente características espaciales relevantes de las imágenes del juego, mientras que la capa totalmente conectada aprende una representación de alto nivel antes de estimar los valores Q asociados a cada acción.

2. Implementación de la solución DQN

````python
def build_agent(nb_actions, window_length, input_shape,
                 double_dqn=False, dueling=False):
    model = build_model(window_length, input_shape, nb_actions, dueling=dueling)

    memory = SequentialMemory(limit=1000000, window_length=window_length)
    processor = AtariProcessor()

    policy = LinearAnnealedPolicy(
        EpsGreedyQPolicy(), attr='eps',
        value_max=1.0, value_min=0.1, value_test=0.05,
        nb_steps=250000
    )

    dqn = DQNAgent(
        model=model, nb_actions=nb_actions, policy=policy, memory=memory,
        processor=processor, nb_steps_warmup=10000, gamma=0.99,
        target_model_update=5000, train_interval=4, delta_clip=1.0,
        enable_double_dqn=double_dqn, enable_dueling_network=dueling,
        dueling_type='avg' if dueling else None
    )
    dqn.compile(Adam(learning_rate=0.00025), metrics=['mae'])
    return dqn

histories = {}
````

### Entrenamiento

````python
from datetime import datetime
from os import name

def train_agent(name):

    # Timestamp único para este arranque de fit()
    run_timestamp = datetime.now().strftime('%Y%m%d_%H%M%S')

    cfg = experiments[name]

    print(f"\n===== Entrenando: {name} a las {run_timestamp} =====")
    env = gym.make(env_name)
    env.seed(123)

    dqn = build_agent(nb_actions, WINDOW_LENGTH, INPUT_SHAPE, **cfg)
    dqn.step = 0

    weights_filename = f'dqn_{env_name}_{name}_weights_{run_timestamp}.h5f'
    checkpoint_filename = f'dqn_{env_name}_{name}_checkpoint_{run_timestamp}.h5f'
    log_filename = f'dqn_{env_name}_{name}_log_{run_timestamp}.json'

    callbacks = [
        ModelIntervalCheckpoint(checkpoint_filename, interval=100000),
        FileLogger(log_filename, interval=10000)
    ]

    history = dqn.fit(env, callbacks=callbacks, nb_steps=NB_STEPS_TRAIN,
                        log_interval=10000, visualize=False)

    dqn.save_weights(weights_filename, overwrite=True)
    histories[name] = log_filename
    env.close()
````

###Seleccion de las variantes

Con el objetivo de comparar distintas estrategias de mejora sobre el algoritmo DQN original, se han seleccionado las siguientes variantes:

**Baseline DQN**, que actúa como modelo de referencia.

**Double DQN**, que incorpora una estrategia para reducir la sobreestimación de los valores Q durante el aprendizaje.

**Dueling DQN**, que modifica la arquitectura de la red neuronal para estimar de forma independiente el valor del estado y la ventaja asociada a cada acción.

Estas tres variantes representan algunas de las mejoras más relevantes propuestas sobre DQN y permiten analizar distintos aspectos del proceso de aprendizaje. La versión Baseline sirve como punto de comparación, mientras que Double DQN busca obtener estimaciones más precisas de los valores Q. Por su parte, Dueling DQN resulta especialmente interesante en un entorno como Enduro, donde durante gran parte de la partida muchas acciones producen efectos similares. En estas situaciones, separar el valor del estado de la ventaja de cada acción puede facilitar el aprendizaje y mejorar el rendimiento del agente.

### Testing

````python
import json

run_timestamps = {
    'baseline_dqn':   "20260725_164222",
    'double_dqn':     "20260731_234157",
    'dueling_dqn':    "20260728_142544",
}

def test_agent(name):
    results = {}

    cfg = experiments[name]
    env = gym.make(env_name)
    dqn = build_agent(nb_actions, WINDOW_LENGTH, INPUT_SHAPE, **cfg)

    run_timestamp = run_timestamps.get(name)
    weights_filename = f'dqn_{env_name}_{name}_weights_{run_timestamp}.h5f'
    dqn.load_weights(weights_filename)

    test_history = dqn.test(env, nb_episodes=100, visualize=True)
    rewards = test_history.history['episode_reward']
    nb_steps = test_history.history['nb_steps']
    results[name] = rewards
    mean_reward = np.mean(rewards)
    pct_above_threshold = np.mean(np.array(rewards) >= REWARD_THRESHOLD) * 100

    print(f"{name}: media = {mean_reward:.2f} | "
          f"% episodios ≥ {REWARD_THRESHOLD} pts = {pct_above_threshold:.1f}%")

    env.close()

    # 'rewards' es el array que ya generaste con test_history.history['episode_reward']
    test_results = {
        name: {
            'nb_steps': nb_steps.tolist() if hasattr(nb_steps, 'tolist') else list(nb_steps),
            'rewards': rewards.tolist() if hasattr(rewards, 'tolist') else list(rewards),
            'threshold': REWARD_THRESHOLD,
            'mean': mean_reward,
            'pct_above_threshold': pct_above_threshold,
        }
    }

    with open(f'test_results_{name}_{run_timestamp}.json', 'w') as f:
        json.dump(test_results, f, indent=2)

    print(f"Guardado 'test_results_{name}_{run_timestamp}.json'.")
````

### 3. Justificación de los parámetros seleccionados y de los resultados obtenidos

Con el fin de realizar una comparación justa entre las distintas variantes de DQN, todos los experimentos se han llevado a cabo utilizando la misma configuración de entrenamiento. De esta forma, las diferencias observadas en el rendimiento pueden atribuirse principalmente a las modificaciones introducidas en cada algoritmo y no a cambios en los hiperparámetros.

Los principales hiperparámetros utilizados son los siguientes:

| Hiperparámetro  |         Valor |
| --------------- | ------------: |
| Learning rate   |       0.00025 |
| Gamma           |          0.99 |
| Batch size      |            32 |
| Replay memory   |     1.000.000 |
| Target update   |  10.000 pasos |
| Epsilon inicial |           1.0 |
| Epsilon final   |           0.1 |
| Epsilon decay   | 250.000 pasos |
| Window Length   |             4 |


* **Learning rate (0.00025):** se utiliza una tasa de aprendizaje reducida para favorecer un entrenamiento estable y evitar cambios demasiado bruscos en los pesos de la red neuronal.

* **Factor de descuento ($\gamma = 0.99$):** se emplea un valor elevado para dar importancia a las recompensas futuras, ya que en Enduro muchas decisiones tienen efectos que se reflejan varios pasos más adelante.

* **Tamaño del lote (batch size):** se utilizan minibatches de experiencias almacenadas en el replay buffer para aproximar el gradiente y mejorar la eficiencia del aprendizaje.

* **Replay Memory (1.000.000 de transiciones):** un buffer de gran tamaño permite almacenar experiencias diversas y romper la correlación entre muestras consecutivas, favoreciendo un entrenamiento más estable.

* **Actualización de la red objetivo (cada 10 000 pasos):** la red objetivo se actualiza periódicamente para proporcionar estimaciones más estables durante el aprendizaje y reducir las oscilaciones del entrenamiento.

* **Exploración $\epsilon$-greedy:** la probabilidad de exploración disminuye progresivamente desde un valor inicial de 1 hasta un mínimo de 0.1. Esto permite que el agente explore el entorno durante las primeras etapas del entrenamiento y, posteriormente, aproveche el conocimiento adquirido.

* **Epsilon decay (250 000 pasos):** la reducción gradual de la exploración evita que el agente deje de explorar demasiado pronto y favorece un equilibrio entre exploración y explotación durante el entrenamiento.

* **Window Length (4):** procesar 4 *frames* consecutivos es vital en juegos de conducción. Un solo *frame* estático no revela la velocidad ni la trayectoria de los coches rivales; agruparlos permite a la red neuronal inferir el movimiento y la profundidad espacial para anticiparse y poder esquivarlos a tiempo.

Los resultados obtenidos con el modelo Baseline DQN muestran que el agente es capaz de aprender una estrategia eficaz para desenvolverse en el entorno Enduro. La evolución de las métricas durante el entrenamiento refleja una mejora progresiva del rendimiento y una recompensa media elevada durante la fase de evaluación. Estos resultados convierten al modelo Baseline en una referencia adecuada para comparar el comportamiento del resto de variantes implementadas.


En teoría, Double DQN suele obtener mejores resultados que el DQN clásico al reducir la sobreestimación de los valores Q pero en nuestros experimentos ha ocurrido lo contrario. El agente ha obtenido una recompensa media de aproximadamente 8 puntos, lo que indica que no ha conseguido aprender una estrategia adecuada para jugar a Enduro.

Planteamos como hipótesis que la red ha caído en un mínimo local, adoptando lo que se conoce como una "política subóptima perezosa". En las mecánicas de *Enduro*, acelerar conlleva un riesgo alto de colisionar con otros vehículos y perder puntos. Al parecer, si los pesos de evaluación no se sintonizan bien en las etapas tempranas, el agente descubre rápidamente que la estrategia más "segura" para no recibir penalizaciones es simplemente frenar constantemente o no acelerar nunca. De esta forma, el coche sobrevive sin chocar durante todo el episodio, pero apenas adelanta a los rivales, lo que justifica una puntuación tan baja (8.00) tras 100 episodios de prueba.

Este resultado puede deberse a que la configuración de hiperparámetros que se ha utilizado no era la más adecuada para esta variante o a que el tiempo de entrenamiento no ha sido suficiente para que el agente aprendiera correctamente y lograse salir de esa "zona de confort" inicial.

En el caso de Dueling DQN, los resultados muestran un comportamiento más sólido durante el entrenamiento y una mejora en la recompensa obtenida durante la evaluación respecto al modelo Baseline. Esto indica que la modificación de la arquitectura resulta beneficiosa para este entorno, permitiendo al agente aprender una estrategia más eficaz. En conjunto, esta variante es la que ofrece el mejor rendimiento de las tres evaluadas en este trabajo.

````python
import json
import matplotlib.pyplot as plt

run_timestamps = {
    'baseline_dqn':   "20260725_164222",
    'double_dqn':     "20260731_234157",
    'dueling_dqn':    "20260728_142544",
}

def generate_training_graphs(name):

    with open(f'dqn_Enduro-v0_{name}_log_{run_timestamps[name]}.json') as f:
        log_data = json.load(f)

    print(log_data.keys())
    print("Episodios registrados:", len(log_data['episode_reward']))

    # Eje X: pasos acumulados de entrenamiento (más informativo que el índice de episodio)
    steps = log_data['nb_steps']

    fig, axes = plt.subplots(4, 1, figsize=(10, 14), sharex=True)

    axes[0].plot(steps, log_data['episode_reward'])
    axes[0].set_ylabel('Reward por episodio')
    axes[0].set_title(f'Progresión del entrenamiento — {name}')
    axes[0].grid(True)

    axes[1].plot(steps, log_data['mean_q'], color='orange')
    axes[1].set_ylabel('mean_q')
    axes[1].grid(True)

    axes[2].plot(steps, log_data['mae'], color='green')
    axes[2].set_ylabel('mae')
    axes[2].grid(True)

    axes[3].plot(steps, log_data['loss'], color='red')
    axes[3].set_ylabel('loss')
    axes[3].set_xlabel('Pasos de entrenamiento')
    axes[3].grid(True)

    plt.tight_layout()
    plt.show()
````

````python
import json
import numpy as np
import matplotlib.pyplot as plt


def generate_test_graphs(name):
    with open(f'test_results_{name}_{run_timestamps[name]}.json') as f:
        test_results = json.load(f)

    rewards_arr = np.array(test_results[name]['rewards'])
    threshold = test_results[name]['threshold']

    fig, axes = plt.subplots(1, 2, figsize=(14, 5))

    # Gráfica 1: recompensa por episodio de test, con la línea de umbral
    axes[0].plot(range(1, len(rewards_arr) + 1), rewards_arr, marker='o', markersize=3)
    axes[0].axhline(threshold, color='red', linestyle='--', label=f'Umbral día 1 ({threshold})')
    axes[0].set_xlabel('Episodio de test')
    axes[0].set_ylabel('Recompensa')
    axes[0].set_title(f'{name} — Recompensa en 100 episodios de test')
    axes[0].legend()
    axes[0].grid(True)

    # Gráfica 2: distribución (histograma)
    axes[1].hist(rewards_arr, bins=15, edgecolor='black')
    axes[1].axvline(threshold, color='red', linestyle='--', label=f'Umbral día 1 ({threshold})')
    axes[1].set_xlabel('Recompensa')
    axes[1].set_ylabel('Nº de episodios')
    axes[1].set_title(f'{name} — Distribución de recompensas')
    axes[1].legend()
    axes[1].grid(True)

    plt.tight_layout()
    plt.savefig(f'test_{name}.png', dpi=150)
    plt.show()

    print(f"Media: {test_results[name]['mean']:.2f}")
    print(f"% episodios ≥ {threshold}: {test_results[name]['pct_above_threshold']:.1f}%")
````

#### Gráficas comparativas

````python
import matplotlib.pyplot as plt

# 1. Curva de aprendizaje (recompensa media móvil durante el entrenamiento)
histories = {
    'baseline_dqn': f'dqn_Enduro-v0_baseline_dqn_log_{run_timestamps["baseline_dqn"]}.json',
    'double_dqn': f'dqn_Enduro-v0_double_dqn_log_{run_timestamps["double_dqn"]}.json',
    'dueling_dqn': f'dqn_Enduro-v0_dueling_dqn_log_{run_timestamps["dueling_dqn"]}.json',
}
plt.figure(figsize=(10, 6))
for name, log_file in histories.items():
    with open(log_file) as f:
        log_data = json.load(f)
    rewards = log_data['episode_reward']
    window = 50
    moving_avg = np.convolve(rewards, np.ones(window)/window, mode='valid')
    plt.plot(moving_avg, label=name)
plt.xlabel('Episodio')
plt.ylabel('Recompensa media móvil (ventana=50)')
plt.title('Curva de aprendizaje por propuesta')
plt.legend()
plt.grid(True)
plt.show()

# 2. Distribución de recompensas en test (boxplot)
test_results = {
    'baseline_dqn': f'test_results_baseline_dqn_{run_timestamps["baseline_dqn"]}.json',
    'double_dqn': f'test_results_double_dqn_{run_timestamps["double_dqn"]}.json',
    'dueling_dqn': f'test_results_dueling_dqn_{run_timestamps["dueling_dqn"]}.json',
}
results = {}
for name, test_file in test_results.items():
    with open(test_file) as f:
        test_data = json.load(f)
    results[name] = test_data[name]['rewards']
    REWARD_THRESHOLD = test_data[name]['threshold']
plt.figure(figsize=(8, 6))
plt.boxplot([results[name] for name in results], labels=list(results.keys()))
plt.axhline(REWARD_THRESHOLD, color='red', linestyle='--', label='Objetivo día 1')
plt.ylabel('Recompensa por episodio (test)')
plt.title('Distribución de recompensas en 100 episodios de test')
plt.legend()
plt.grid(True)
plt.show()

# 3. Recompensa por episodio de test (comparación directa)
plt.figure(figsize=(10, 6))
for name in results:
    plt.plot(results[name], label=name, alpha=0.7)
plt.axhline(REWARD_THRESHOLD, color='red', linestyle='--', label='Objetivo día 1')
plt.xlabel('Episodio de test')
plt.ylabel('Recompensa')
plt.title('Recompensa obtenida en cada uno de los 100 episodios de test')
plt.legend()
plt.grid(True)
plt.show()
````

En las gráficas comparativas se pueden ver claramente las diferencias de rendimiento entre las tres variantes implementadas.

Se observa que Baseline DQN y Dueling DQN consiguen aprender una estrategia eficaz para resolver el entorno, mientras que Double DQN no alcanza un rendimiento similar con la configuración utilizada. Entre las tres variantes, Dueling DQN obtiene los mejores resultados, mostrando una evolución más estable durante el entrenamiento y las mayores recompensas en la fase de evaluación.

En conjunto, estos resultados indican que la arquitectura Dueling DQN es la que mejor se adapta al entorno Enduro de las alternativas analizadas.
